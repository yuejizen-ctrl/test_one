```python
#要存到50000元，存款到了後提醒
deposit=0
mounth=0
while deposit<50000:
    mounth+=1
    inputnum=int(input(f"請輸入第{mounth}存錢金額"))
    deposit+=inputnum
print (f"存了{mounth}月{deposit}")
```

```python
deposit=0
mounth=0
while True:
     mounth+=1
     inputnum=int(input(f"請輸入第{mounth}存錢金額"))
     deposit+=inputnum
     if deposit>=50000:
          break
print (f"存了{mounth}月{deposit}")
```

```python
sum=0
num=0
while True:
    value=int(input(f"請輸入一個職{num+1}"))
    if value<0:
        break
    num+=1
    if value %2==1:
        continue
    sum+=value
print(f"輸入的次數{num},偶數加總{sum}")
```
