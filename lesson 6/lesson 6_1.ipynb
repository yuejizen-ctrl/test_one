```python
#猜數字遊戲
import random

min=1
max=100
count=0

target=random.randint(min,max)
print(target)
print("===========猜數字遊戲==========: \n")


while(True):
    count+=1
    keyin=int(input(f"請輸入數字，並在{min}~{max}範圍內"))
    if keyin>=min and keyin<=max:
        if keyin ==target:
            print(f"你猜中了,答案是{target}")
            print(f"你共猜了{count}次")
            break
        else:
            print(f"你猜錯了")
            print(f"你共猜了{count}次")
    else:
        print("不再範圍內")
        break
print("遊戲結束")
```
